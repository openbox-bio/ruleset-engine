**<center>Instructions on maintaining ruleset.</center>**

Use the fernet module from cryptography to generate a key.
<br>`from cryptography.fernet import Fernet`
<br>`key = Fernet.generate_key()`
<br> This key can then be used to encrypt the grammar file. 
<br> Be sure to update this key in the ruleset-engine code and recreate the executable using Pyinstaller.

Here is the Fernet key used to encrypt the grammar file.
<br> **Note: if you update the Fernet key, be sure to update ruleset-engine.py to update the stored Fernet key in that program.**

QGQ8NXw8hOgMoCe4XZ3PR7W95JC86qvP-jFo9v5QvY8=

<b>If you update any part of the code: the ruleset modules or the ruleset engine, then the ruleset-engine executable will have to be regenerated using pyinstaller:</b>
<br> `pyinstaller --onefile ruleset-engine.py`

<b>if you update *Ruleset.tx* then you have to recreate the encrypted file *Ruleset.tx.enc*.</b>
<br> See below for code to create the \*.enc file.

```python
from cryptography.fernet import Fernet
```

```python
key = 'QGQ8NXw8hOgMoCe4XZ3PR7W95JC86qvP-jFo9v5QvY8='
f = Fernet(key)
```

```python
with open('RuleSet.tx', 'rb') as file:
    original = file.read()
```

```python
encrypted = f.encrypt(original)
```

```python
with open('RuleSet.tx.enc', 'wb') as encrypted_file:
    encrypted_file.write(encrypted)
```
